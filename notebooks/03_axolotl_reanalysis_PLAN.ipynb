## 1. Biological Objective

The axolotl dataset will be analyzed to identify regeneration-associated cellular states, marker genes, and differentially expressed genes that can later be compared with mouse regenerative digit-tip data and the human fingertip regeneration reference.

The analysis should focus on biologically relevant regeneration-associated populations rather than attempting to combine all 1,604 GEO sample records into one matrix.

## 3. Published QC Criteria

The original GSE106269 study reported the following QC/data-processing criteria:

1. Exclude cells with **fewer than 1,000 detected genes**.
2. Exclude **ribosomal protein genes**.
3. For limb-bud samples, remove cells with **Prrx1 = 0** as an identity check.
4. No mitochondrial-percentage cutoff was reported.
5. The study used a curated axolotl transcriptome because genome annotation was incomplete.

### Planned implementation

The Scanpy workflow will reproduce the reported criteria where applicable.

Mitochondrial percentage will be **calculated and inspected as a QC metric**, but no arbitrary mitochondrial cutoff will be imposed before examining the actual dataset.

## 6. Regeneration-Focused Analysis

The original study reported convergence of differentiated connective-tissue populations toward a shared progenitor-like state following amputation, followed by later re-diversification.

The reanalysis should therefore examine:

- Cell-type composition across regeneration time points
- Connective-tissue populations
- Cycling/progenitor-associated states
- Blastema-associated transcriptional signatures
- Changes in marker expression across time
- Cluster-specific differential expression

The analysis should preserve the distinction between:

**cell identity → cell state → regeneration-associated signature**

A regeneration-associated signature should not automatically be described as a unique regeneration-specific gene.


## 7. Differential Expression Plan

After clustering and annotation:

- Identify marker genes for relevant clusters.
- Compare biologically relevant cell populations/states.
- Where appropriate, compare regeneration time points or states.
- Record statistical method and thresholds in the final analysis notebook.
- Generate ranked gene lists for downstream ortholog mapping.

Only genes supported by the defined statistical and biological criteria will proceed to cross-species comparison.

## 9. Computational Constraints

The available system has limited RAM for very large single-cell matrices.

Therefore:

- Do not load all GSE106269 subsets simultaneously unless required.
- Prefer the biologically relevant processed subset.
- Avoid unnecessary conversion of large matrices.
- Save intermediate results only when useful.
- Monitor memory usage during large operations.
- Use sparse matrices where appropriate.
- Do not generate a large `.h5ad` file containing unnecessary datasets.

The previous Jupyter/WSL instability during large matrix processing makes memory-aware analysis important.
