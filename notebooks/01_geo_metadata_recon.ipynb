```python
for key, value in gsm.metadata.items():
    if key in ['title', 'source_name_ch1', 'characteristics_ch1',
               'platform_id', 'instrument_model', 'library_strategy',
               'supplementary_file_1']:
        print(f"\n{key}:")
        print(value)
```

```python
from collections import Counter

stage_counts = Counter()

for gsm in gse_axolotl.gsms.values():
    for characteristic in gsm.metadata.get('characteristics_ch1', []):
        stage_counts[characteristic] += 1

for characteristic, count in stage_counts.most_common():
    print(count, "-", characteristic)
```

```python
from collections import Counter

platform_counts = Counter()

for gsm in gse_axolotl.gsms.values():
    platform = gsm.metadata.get('platform_id', ['Unknown'])[0]
    platform_counts[platform] += 1

for platform, count in platform_counts.most_common():
    print(count, "-", platform)
```

### GSE106269 — Axolotl

GSE106269 contains 1,604 GEO sample records from the axolotl limb regeneration study by Gerber et al. The dataset contains multiple single-cell RNA-seq experimental subsets and platforms, including Fluidigm C1, SmartSeq2, and 10X Genomics. Sample metadata identify developmental limb-bud stages, uninjured tissue, and multiple post-amputation blastema timepoints. Series-level supplementary files include the raw data archive (`GSE106269_RAW.tar`) and several supplementary study tables, while individual 10X samples have sample-level barcode files. The dataset will therefore require careful subset selection and metadata-based identification before downstream scRNA-seq analysis.

### GSE131078 — Mouse

GSE131078 contains 24 bulk RNA-seq samples from mouse hind-limb digit tips, consisting of 12 regenerative (distal amputation) and 12 non-regenerative (proximal amputation) samples. Each condition contains four biological replicates at 12, 14, and 21 days post-amputation (DPA), giving a balanced 2 × 3 × 4 experimental structure. GEO metadata clearly identify treatment, timepoint, tissue, strain, and sequencing information. A single series-level processed gene-count matrix (`GSE131078_gene_counts.txt.gz`) is available and no individual sample supplementary files are listed. This processed matrix will be used for downstream bulk RNA-seq analysis after its structure and gene/sample identifiers are verified.
